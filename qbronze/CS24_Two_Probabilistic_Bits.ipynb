Suppose that we have two probabilistic bits, and our probabilistic states respectively are

$ \begin{pmatrix} 0.2 \\ 0.8 \end{pmatrix} \quad{ and } \begin{pmatrix} 0.6 \\ 0.4 \end{pmatrix}. $

If we combine both bits as a single system, then what is the state of the combined system?

In total, we have four different states. We can name them as follows:
<ul>
    <li>00: both bits are in states 0</li>
    <li>01: the first bit is in state 0 and the second bit is in state 1</li>
    <li>10: the first bit is in state 1 and the second bit is in state 0</li>
    <li>11: both bits are in states 1</li>
</ul>

<h3> Task 1 </h3>

<b>Discussion and analysis:</b>

What are the probabilities of being in states $ 00 $, $ 01 $, $ 10 $, and $11$?

How can we represent these probabilities as a column vector?

<h3> Representation for states 0 and 1</h3>

The vector representation of state 0 is $ \begin{pmatrix} 1 \\ 0 \end{pmatrix}$. Similarly, the vector representation of state 1 is $ \begin{pmatrix} 0 \\ 1 \end{pmatrix} $.

We use $\lceil 0 \rfloor$  to represent $ \begin{pmatrix} 1 \\ 0 \end{pmatrix} $ and $\lceil 1 \rfloor$   to represent $\begin{pmatrix} 0 \\ 1 \end{pmatrix}$.

Then, the probabilistic state $ \begin{pmatrix} 0.2 \\ 0.8 \end{pmatrix} $ is also represented as $ 0.2 \lceil 0 \rfloor + 0.8 \lceil 1 \rfloor $.

Similarly, the probabilistic state $ \begin{pmatrix} 0.6 \\ 0.4 \end{pmatrix} $ is also represented as $ 0.6 \lceil 0 \rfloor + 0.4 \lceil 1 \rfloor $.

<h3> Composite systems </h3>

When two systems are composed, then their states are tensored to calculate the state of composite system.

The probabilistic state of the first bit is $  \begin{pmatrix} 0.2 \\ 0.8 \end{pmatrix} = $ $0.2 \lceil 0 \rfloor + 0.8 \lceil 1 \rfloor $.

The probabilistic state of the second bit is $  \begin{pmatrix} 0.6 \\ 0.4 \end{pmatrix} = 0.6 \lceil 0 \rfloor + 0.4 \lceil 1 \rfloor $.

Then, the probabilistic state of the composite system is $ \big( 0.2 \lceil 0 \rfloor + 0.8 \lceil 1 \rfloor \big) \otimes  \big( 0.6 \lceil 0 \rfloor + 0.4 \lceil 1 \rfloor \big) $.


<h3> Task 2 </h3>

Find the probabilistic state of the composite system.

<i> 
Rule 1: Tensor product distributes over addition in the same way as the distribution of multiplication over addition.

Rule 2: $ \big( 0.3 \lceil 1 \rfloor \big) \otimes \big( 0.7 \lceil 0 \rfloor \big) = (0.3 \cdot 0.7) \big( \lceil 1 \rfloor \otimes \lceil 0 \rfloor \big) = 0.21 \lceil 10 \rfloor $.
</i>

<h3> Extra: Task 5 </h3>

Suppose that we have three bits.

Find the vector representations of $ \lceil abc \rfloor $ for each $ a,b,c \in \{0,1\} $.

```python
[a1b1c1, a1b1c2, a1b2c1, a1b2c2, a2b1c1, a2b1c2, a2b2c1, a2b2c2]
```

```python
def bit_vector(bit):
	return [1, 0] if bit == 0 else [0, 1]

def tensor(v1, v2):
	return [x * y for x in v1 for y in v2]

for a in (0, 1):
	for b in (0, 1):
		for c in (0, 1):
			vec = tensor(tensor(bit_vector(a), bit_vector(b)), bit_vector(c))
			print(f"{a}{b}{c}:", vec)
```
